```python
import pandas as pd
from sklearn.datasets import load_breast_cancer,fetch_california_housing
from sklearn.ensemble import RandomForestClassifier,RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix,precision_score,recall_score,accuracy_score,r2_score,mean_absolute_error
```

```python
cdata =load_breast_cancer()
df = pd.DataFrame(cdata.data,columns=cdata.feature_names)
df['cancer'] = cdata.target
df.head()
```

```python
X = df.drop("cancer", axis=1)
y = df['cancer']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
cmod = RandomForestClassifier(n_estimators=200,max_depth=10)
cmod.fit(X_train,y_train)
```

```python
ypred = cmod.predict(X_test)
ypred
```

```python
print("Training Accuracy:", cmod.score(X_train, y_train))
print("Testing Accuracy:", cmod.score(X_test, y_test))
```

```python
rdata = fetch_california_housing()
rdf = pd.DataFrame(rdata.data,columns=rdata.feature_names)
rdf['target'] = rdata.target
rdf.head()
```

```python
X = rdf.drop('target', axis=1)
y = rdf['target']
```

```python
Xtrain,Xtest,ytrain,ytest = train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
rmod = RandomForestRegressor(n_estimators=150,max_depth=10)
rmod.fit(Xtrain,ytrain)
```

```python
ypred =rmod.predict(Xtest)
ypred
```

```python
print('Training Accuracy: ',rmod.score(Xtrain,ytrain))
print('Testing Accuracy:', rmod.score(Xtest,ytest))
```

```python
r2_score(ytest, ypred)
```

```python
mean_absolute_error(ytest, ypred)
```
