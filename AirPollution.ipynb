```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('/content/Measurement_info.csv')
data
item = pd.read_csv('/content/Measurement_item_info.csv')
item
station = pd.read_csv('/content/Measurement_station_info.csv')
station
```

# New Section

```python

sub_station = station[['Station code', 'Station name(district)']].drop_duplicates()
sub_station

```

```python
data_i = data.merge(item[['Item code', 'Item name']])
data_i
```

```python
data_s = data_i.merge(
    station[['Station code', 'Station name(district)']]
)
data_s
```

```python
data = data_s.drop(columns=['Station code', 'Item code'])
data.head()
```

```python
# Dictionary already given
status_dict = {
    0: "Normal",
    1: "Need for calibration",
    2: "Abnormal",
    4: "Power cut off",
    8: "Under repair",
    9: "Abnormal data"
}

# Convert dictionary to DataFrame
dictdf = pd.DataFrame(list(status_dict.items()),
                         columns=['Instrument status', 'Instrument Status Name'])

# Merge with data
data = data.merge(dictdf, on='Instrument status', how='left')

# Drop the old column
data = data.drop(columns=['Instrument status'])

data.head()
```

```python
# Convert Measurement date to datetime
data['Measurement date'] = pd.to_datetime(data['Measurement date'])

# Extract year, month, day, hour
data['Year'] = data['Measurement date'].dt.year
data['Month'] = data['Measurement date'].dt.month
data['Day'] = data['Measurement date'].dt.day
data['Hour'] = data['Measurement date'].dt.hour

# Drop the original column
data = data.drop(columns=['Measurement date'])

data.head()
```
