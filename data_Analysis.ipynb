```python
import re

# Function to check for special characters in names
def has_special_characters(name):
    if pd.isnull(name):
        return False
    return bool(re.search(r'[^a-zA-Z\s]', name))

special_char_names = df[df['Name'].apply(has_special_characters)]
special_char_names.head(20)
```

```python
def contains_letters(age_value):
    if age_value.lower() == 'nan':  # skip nan
        return False
    return any(char.isalpha() for char in age_value) # True if any character is a letter
```
