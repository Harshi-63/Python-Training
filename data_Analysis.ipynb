```python
salary_series = df['Salary']
salary_str = salary_series.astype(str).str.strip()
unique_salaries = salary_str.unique()
```

```python


def salary_format_type(s):
    s = str(s).strip()

    if s.lower() == 'nan':
        return 'Missing'

    elif any(c.isalpha() for c in s):
        return 'Contains letters (k, M, etc.)'

    elif any(c in '₹$,' for c in s):
        return 'Currency/Comma'

    elif s.replace('.', '', 1).isdigit():
        return 'Numeric'

    else:
        return 'Other/Invalid'


# Apply on Salary column
Salary_format = df['Salary'].apply(salary_format_type)

# Count each type
Salary_format.value_counts()
```
